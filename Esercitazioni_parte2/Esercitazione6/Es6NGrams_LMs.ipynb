```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.lm.preprocessing import padded_everygram_pipeline
from nltk.lm import MLE
from nltk.corpus import gutenberg

nltk.download('gutenberg')
nltk.download('punkt')
nltk.download('punkt_tab')

text_politico = [
   "We must build a better future for our great country.",
    "The economy is growing, and jobs are finally coming back.",
    "Vote for progress, vote for a better tomorrow.",
    "Taxes must be lowered to help the working middle class.",
    "Our healthcare system needs a complete and fair reform.",
    "Education is the cornerstone of our great nation's success.",
    "Together we stand strong against those who want to divide us.",
    "I will never stop fighting for the hardworking families of this country.",
    "Infrastructure investments will bring millions of new well-paying jobs.",
    "We must protect our borders and ensure our national security.",
    "Clean energy is the true future of our economy and our planet.",
    "Freedom and liberty are the core values we must defend every single day.",
    "A strong and wealthy middle class means a strong and healthy democracy.",
    "Let us unite today to overcome the difficult challenges of this generation.",
    "No citizen should have to choose between buying food and buying medicine."
]

text_tech = [
   "AI and machine learning are going to change the world forever.",
    "Just launched our new crypto startup on the blockchain today.",
    "Data science and artificial intelligence are the true future of tech.",
    "Disrupting the entire industry with our new autonomous mobile app.",
    "Coding all night in Python to push the new software update.",
    "Coffee, code, deploy, repeat. The startup hustle never ever stops.",
    "Just secured our Series A funding! Time to scale the team 10x.",
    "Web3 is fundamentally shifting how we think about digital ownership.",
    "If you are not automating your workflows, you are falling way behind.",
    "Our new neural network architecture achieved state of the art results.",
    "Super excited to announce our integration with the new cloud API.",
    "Move fast and break things. That is the only real way to innovate.",
    "Virtual reality and augmented reality will redefine human interaction.",
    "Debugging is like being the detective in a movie where you are the murderer.",
    "Building the decentralized metaverse one line of code at a time."
]
```

```python

def train_and_generate(corpus_sentences, n_gram_size, num_words=20):
    #NLTK: n-grammi e vocabolario
    train_data, padded_vocab = padded_everygram_pipeline(n_gram_size, corpus_sentences)

    #addestramento
    model = MLE(n_gram_size)
    model.fit(train_data, padded_vocab)
    generated_tokens = model.generate(num_words)

    # Pulizia
    clean_text = []
    for token in generated_tokens:
        if token == '<s>': continue
        if token == '</s>': break
        clean_text.append(token)

    return " ".join(clean_text)
```

```python
# Pre-processing dei tweet
text_politico = [word_tokenize(t.lower()) for t in text_politico]
text_tech = [word_tokenize(t.lower()) for t in text_tech]

print("\n> Stile POLITICO:")
print(" Bigrammi :", train_and_generate(text_politico, n_gram_size=2, num_words=15))
print(" Trigrammi:", train_and_generate(text_politico, n_gram_size=3, num_words=15))

print("\n> Stile TECH:")
print(" Bigrammi :", train_and_generate(text_tech, n_gram_size=2, num_words=15))
print(" Trigrammi:", train_and_generate(text_tech, n_gram_size=3, num_words=15))

```

**Si nota come i trigrammi del modello politico producano frasi molto più coerenti rispetto ai bigrammi, mantenendo intatto lo stile propagandistico**

```python
print("\n GENERAZIONE MOBY DICK")
print("Caricamento corpus Moby Dick")

# NLTK fornisce Moby Dick già diviso in frasi e tokenizzato
moby_dick_sents = gutenberg.sents('melville-moby_dick.txt')

# passiamo tutte le frasi al modello
# Convertiamo in minuscolo per ridurre la sparsità del vocabolario
moby_dick_lower = [[word.lower() for word in sent] for sent in moby_dick_sents]

print("\n> Moby Dick - BIGRAMMI (Contesto corto, più caotico):")
for i in range(5):
    print(f" {i+1}.", train_and_generate(moby_dick_lower, n_gram_size=2, num_words=20))

print("\n> Moby Dick - TRIGRAMMI (Contesto lungo, più strutturato):")
for i in range(5):
    print(f" {i+1}.", train_and_generate(moby_dick_lower, n_gram_size=3, num_words=20))
```
